# 00 · Setup

**Goal:** load the smallest instruct model we can find and get one reply out of it. About two minutes on a laptop CPU, no GPU needed.

We use **SmolLM2-135M-Instruct** (135 million parameters). It is far too small to be a good assistant — that's fine. Everything you learn here (activations, probes, steering) works the same way on bigger models; you just swap one string, `MODEL_ID` in `lm.py`.

```python
# On Colab, uncomment and run these three lines first:
# !git clone <repo-url>
# %cd rig-interpretability-hackathon/starter/notebooks
# !pip install -q -r ../requirements.txt

import sys; sys.path.insert(0, "..")      # so the helper files in starter/ import
import numpy as np
import matplotlib.pyplot as plt
```

```python
from lm import load_model, describe, chat, MODEL_ID

print("loading", MODEL_ID)
model, tok = load_model()
describe(model)
```

`layers` is how many transformer blocks the model has. In this repo, **layer 0 is the token embeddings and layer *k* is the residual stream after block *k***, so there are `layers + 1` places you can read activations from. `d_model` is the width of each activation vector.

```python
chat(model, tok, "In one sentence, what is a sandbox in computer security?")
```

## Tokens

The model never sees text, only tokens. Activations exist *per token*. Throughout the starter we read the activation at the **last token** of each input — by then the model has read the whole thing.

```python
ids = tok("The billing service is returning intermittent errors.")["input_ids"]
print(len(ids), "tokens:", tok.convert_ids_to_tokens(ids))
```
